```python
import numpy as np

i_hat = np.array([1,0])
j_hat = np.array([0,1])

vector = 2*i_hat + 3* j_hat
print(vector)
```

```python
x = np.array([2, 3])
scale = np.array([[2, 0], [0, 3]])
a = scale @ x
print(a)
```

```python
a = np.array([[2, 0], [0, 3]])
b = np.array([[3, 4], [5, 6]])
x = np.array([1,0])
result = a @ (b@x)
result
result2 = (x@b) @ a
result2
```

```python
np.linalg.det(a)
```

```python
np.linalg.inv(a)
```

```python
np.linalg.solve(a, b)
```

```python
np.linalg.matrix_rank(a)
```
